# IBM Quantum setup (run once)
Connects this environment to IBM Quantum so section 7b of `quantum_rps.ipynb` can run on real hardware.

- `getpass` asks for your free IBM Quantum API key in a hidden box, so the key is **never shown or saved in this notebook**.
- `save_account` stores it privately in your own environment (`~/.qiskit`), not in the project files.
- The output lists the real IBM quantum computers your account can use. Mine were ibm_fez, ibm_kingston and ibm_marrakesh (156 qubits each).

⚠️ Never paste your API key directly into a code cell.

```python
from getpass import getpass
from qiskit_ibm_runtime import QiskitRuntimeService

api_key = getpass("Paste your IBM API key and press Enter: ")   # hidden: never shown or saved in the notebook

QiskitRuntimeService.save_account(token=api_key, set_as_default=True, overwrite=True)

service = QiskitRuntimeService()
for backend in service.backends():
    print(backend.name, "-", backend.num_qubits, "qubits")
```
